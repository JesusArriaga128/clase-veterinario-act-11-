```python
# --- PROGRAMA: ASISTENTE VETERINARIO INTELIGENTE ---

# 1. Inicializamos la lista de diccionarios vacía
lista_veterinarios = []

# 2. Definición de Funciones

def registrar_mascota():
    print("\n--- REGISTRO DE NUEVA MASCOTA ---")
    nombre = input("Nombre de la mascota: ")
    especie = input("Especie (Perro / Gato / Ave): ").lower()
    edad = input("Edad de la mascota: ")
    sintoma = input("Síntoma principal (Fiebre / Tos / Alergia / Ninguno): ").lower()

    # Simulación de Inteligencia Artificial (Sistema Experto basado en reglas)
    # La IA analiza los síntomas y asigna un diagnóstico y nivel de urgencia automáticamente
    if sintoma == "fiebre":
        diagnostico = "Posible infección (Requiere revisión)"
        prioridad = "ALTA"
    elif sintoma == "tos":
        diagnostico = "Problema respiratorio ligero"
        prioridad = "MEDIA"
    elif sintoma == "alergia":
        diagnostico = "Reacción cutánea o alimentaria"
        prioridad = "BAJA"
    else:
        diagnostico = "Chequeo de rutina / Saludable"
        prioridad = "STABLE"

    # Creamos el diccionario con los atributos (Keys y Values)
    mascota = {
        "Nombre": nombre,
        "Especie": especie.capitalize(),
        "Edad": edad,
        "Síntoma": sintoma.capitalize(),
        "Diagnóstico IA": diagnostico,
        "Prioridad": prioridad
    }

    # Agregamos el diccionario a nuestra lista
    lista_veterinarios.append(mascota)
    print(f"¡Mascota {nombre} registrada y analizada por la IA con éxito!")


def mostrar_tabla_pacientes():
    print("\n==================================== TABLA DE PACIENTES VETERINARIOS ====================================")
    # Imprimimos los encabezados con un formato alineado usando (.ljust) para simular una tabla
    print("Nombre".ljust(15) + "Especie".ljust(12) + "Edad".ljust(8) + "Síntoma".ljust(15) + "Diagnóstico IA".ljust(35) + "Prioridad")
    print("-" * 105)

    # Ciclo FOR para recorrer la lista de diccionarios
    for mascota in lista_veterinarios:
        print(
            mascota["Nombre"].ljust(15) +
            mascota["Especie"].ljust(12) +
            mascota["Edad"].ljust(8) +
            mascota["Síntoma"].ljust(15) +
            mascota["Diagnóstico IA"].ljust(35) +
            mascota["Prioridad"]
        )
    print("=========================================================================================================\n")


def consultar_ia():
    print("\n--- CONSULTOR DE SÍNTOMAS IA ---")
    print("Pregúntale a la IA qué hacer según el síntoma de un animal.")
    buscar_sintoma = input("Escribe el síntoma a consultar (Fiebre, Tos, Alergia): ").lower()

    # Estructuras condicionales para las respuestas de la IA
    if buscar_sintoma == "fiebre":
        print("🤖 IA responde: Mantén a la mascota hidratada y agenda una cita urgente. No le des medicamentos humanos.")
    elif buscar_sintoma == "tos":
        print("🤖 IA responde: Evita cambios bruscos de temperatura. Si persiste por más de 2 días, traer a revisión.")
    elif buscar_sintoma == "alergia":
        print("🤖 IA responde: Revisa si cambiaste su alimento recientemente. Baña a la mascota con champú neutro.")
    else:
        print("🤖 IA responde: Síntoma no reconocido en la base de datos actual. Se recomienda observación.")


# 3. Menú Interactivo Principal

while True:
    print("==========================================")
    print("   SISTEMA IA - ASISTENTE VETERINARIO     ")
    print("==========================================")
    print("1. Registrar nueva mascota y diagnosticar")
    print("2. Mostrar tabla de pacientes actuales")
    print("3. Consultar recomendaciones de la IA")
    print("4. Salir del programa")
    print("==========================================")

    opcion = input("Selecciona una opción (1-4): ")

    if opcion == "1":
        registrar_mascota()
    elif opcion == "2":
        # Verificamos si la lista está vacía antes de mostrar la tabla
        if len(lista_veterinarios) == 0:
            print("\n[!] No hay pacientes registrados todavía.")
        else:
            mostrar_tabla_pacientes()
    elif opcion == "3":
        consultar_ia()
    elif opcion == "4":
        print("\n¡Gracias por usar el Asistente Veterinario IA! Saliendo del sistema...")
        break
    else:
        print("\n[Opción Inválida] Por favor, elige un número del 1 al 4.")
```
