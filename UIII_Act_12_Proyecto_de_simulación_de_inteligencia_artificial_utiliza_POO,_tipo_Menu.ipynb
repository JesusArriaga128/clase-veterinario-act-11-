```python
# =========================================================================
# PLANTEL: CBTIS No. 128
# SUBMÓDULO 2: Elabora proyectos de programación lógica
# PROFESOR: Eliseo Nava (Especialidad: Inteligencia Artificial)
# ALUMNO: Jesus Arriaga
# =========================================================================

# --- CLASE 1: EL OBJETO VETERINARIO ---
class Veterinario:
    def __init__(self, nombre, especialidad, clinica, cirugias_realizadas, vacunas_disponibles, licencia, horario_urgencias):
        self.nombre = nombre
        self.especialidad = especialidad.lower()
        self.clinica = clinica
        self.cirugias_realizadas = int(cirugias_realizadas)
        self.vacunas_disponibles = int(vacunas_disponibles)
        self.licencia = licencia
        self.horario_urgencias = horario_urgencias.lower()

    # Función interna: Simulación de IA para evaluar aptitud ante un caso
    def simular_ia_evaluacion(self, tipo_mascota, requiere_cirugia, requiere_vacuna, es_urgencia_nocturna):
        puntuacion_aptitud = 0

        if tipo_mascota.lower() in self.especialidad or self.especialidad == "general":
            puntuacion_aptitud += 40

        if requiere_cirugia == "si":
            if self.cirugias_realizadas > 50:
                puntuacion_aptitud += 30
            else:
                puntuacion_aptitud += 10
        else:
            puntuacion_aptitud += 20

        if requiere_vacuna == "si":
            if self.vacunas_disponibles > 0:
                puntuacion_aptitud += 20
            else:
                puntuacion_aptitud -= 10
        else:
            puntuacion_aptitud += 10

        if es_urgencia_nocturna == "si":
            if "24" in self.horario_urgencias or "nocturno" in self.horario_urgencias:
                puntuacion_aptitud += 10
            else:
                puntuacion_aptitud -= 30

        if puntuacion_aptitud >= 80:
            return "RECOMENDADO (Apto para atención inmediata)"
        elif puntuacion_aptitud >= 50:
            return "ACEPTABLE (Puede atender con limitaciones)"
        else:
            return "NO RECOMENDADO (Falta especialidad o infraestructura)"


# --- CLASE 2: EL GESTOR DEL SISTEMA (POO TOTAL) ---
class SistemaIAVeterinaria:
    def __init__(self):
        self.base_veterinarios = [
            Veterinario("Dr. Carlos Gómez", "Caninos y Felinos", "Animal Health", "120", "25", "REG-5543", "24 Horas"),
            Veterinario("Dra. Laura Flores", "Aves y Exóticos", "Wild Vet", "35", "5", "REG-9921", "Diurno"),
            Veterinario("Dr. Marcos Marín", "General", "Pet Center", "200", "0", "REG-1122", "Nocturno")
        ]

    # FUNCIÓN MODIFICADA: Generar y mostrar la tabla con más espacio interlineal y de columnas
    def mostrar_tabla_veterinarios(self):
        if not self.base_veterinarios:
            print("\n[!] No hay veterinarios registrados en el sistema.")
            return

        # Incrementamos el ancho de las columnas para dar más espacio horizontal
        # Total de ancho de línea = 145 caracteres
        print("\n" + "=" * 145)
        print(f"{'VETERINARIO':<25} | {'ESPECIALIDAD':<20} | {'CLÍNICA':<20} | {'CIRUGÍAS':<12} | {'VACUNAS':<10} | {'LICENCIA':<15} | {'URGENCIAS':<20}")
        print("=" * 145)

        for v in self.base_veterinarios:
            # Imprime los datos alineados en sus columnas ampliadas
            print(f"{v.nombre:<25} | {v.especialidad:<20} | {v.clinica:<20} | {v.cirugias_realizadas:<12} | {v.vacunas_disponibles:<10} | {v.licencia:<15} | {v.horario_urgencias:<20}")
            # Agrega una línea en blanco de separación entre cada veterinario para dar espacio vertical
            print("-" * 145)

        print("\n")

    def consultar_ia_sistema(self):
        if not self.base_veterinarios:
            print("\n[!] No hay veterinarios cargados en el sistema.\n")
            return

        print("\n--- CASO MÉDICO (CONSULTA IA DE ASIGNACIÓN) ---")
        tipo_mascota = input("¿Qué tipo de mascota es? (Perro, Gato, Ave, Hamster): ")
        req_cirugia = input("¿El caso requiere cirugía? (si / no): ").lower()
        req_vacuna = input("¿El motivo es para aplicación de vacuna? (si / no): ").lower()
        es_nocturna = input("¿La emergencia ocurre a altas horas de la noche? (si / no): ").lower()

        print("\n" + "~" * 85)
        print("  PROCESANDO DIAGNÓSTICO DE ASIGNACIÓN CON INTELIGENCIA ARTIFICIAL... ")
        print("~" * 85)
        print(f"{'VETERINARIO':<25} | {'EVALUACIÓN Y DICTAMEN DE LA IA'}")
        print("-" * 85)

        for v in self.base_veterinarios:
            dictamen_ia = v.simular_ia_evaluacion(tipo_mascota, req_cirugia, req_vacuna, es_nocturna)
            print(f"{v.nombre:<25} | {dictamen_ia}")
        print("~" * 85 + "\n")

    def ejecutar_menu(self):
        print("-----------------------------------------------------------------")
        print("  CBTIS 128 - SUBMÓDULO 2: PROYECTO IA VETERINARIA")
        print("  Alumno: Jesus Arriaga  |  Profesor: Eliseo Nava")
        print("-----------------------------------------------------------------")

        menu_activo = True
        while menu_activo:
            print("--- MENÚ DE CONTROL ---")
            print("1. Registrar un nuevo Veterinario")
            print("2. Mostrar plantilla de Veterinarios (Tabla)")
            print("3. Consultar IA (Buscar el mejor veterinario para un caso)")
            print("4. Salir del programa")

            opcion = input("Selecciona una opción (1-4): ")

            if opcion == "1":
                print("\n--- REGISTRO DE NUEVO VETERINARIO ---")
                nombre = input("Nombre del Médico: ")
                especialidad = input("Especialidad (ej. Caninos, Exóticos, General): ")
                clinica = input("Nombre de la Clínica: ")
                cirugias = input("Número de cirugías realizadas: ")
                vacunas = input("Cantidad de vacunas disponibles: ")
                licencia = input("Número de Licencia Médica: ")
                horario = input("Horario de urgencias (ej. 24 Horas, Diurno, Nocturno): ")

                nuevo_vet = Veterinario(nombre, especialidad, clinica, cirugias, vacunas, licencia, horario)
                self.base_veterinarios.append(nuevo_vet)
                print("\n[✔] Veterinario registrado y guardado en la estructura de datos.\n")

            elif opcion == "2":
                self.mostrar_tabla_veterinarios()

            elif opcion == "3":
                self.consultar_ia_sistema()

            elif opcion == "4":
                print("\nSistema finalizado correctamente. ¡Éxito en la presentación, Jesús!")
                menu_activo = False
            else:
                print("\n[!] Opción inválida. Intenta de nuevo.\n")


# --- EJECUCIÓN DEL PROGRAMA ---
programa = SistemaIAVeterinaria()
programa.ejecutar_menu()
```
