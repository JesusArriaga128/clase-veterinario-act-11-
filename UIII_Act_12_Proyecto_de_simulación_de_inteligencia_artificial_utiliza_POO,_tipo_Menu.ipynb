```python
# --- CLASE VETERINARIO (POO) ---
class Veterinario:
    # Atributos de la clase (Inicializador)
    def __init__(self, nombre_mascota, especie, edad, sintoma):
        self.nombre_mascota = nombre_mascota
        self.especie = especie.lower()
        self.edad = edad
        self.sintoma = sintoma.lower()
        self.diagnostico = "No determinado"
        self.tratamiento = "Requiere revisión presencial"

    # Función: Simulación de Inteligencia Artificial para diagnosticar
    def simular_ia_diagnostico(self):
        # El "cerebro" de nuestra IA: un sistema de reglas basado en síntomas
        if "vomito" in self.sintoma or "diarrea" in self.sintoma:
            self.diagnostico = "Infección estomacal o parásitos"
            self.tratamiento = "Dieta blanda, hidratación y antibiótico veterinario bajo receta."

        elif "estornudo" in self.sintoma or "tos" in self.sintoma:
            self.diagnostico = "Resfriado o infección respiratoria"
            self.tratamiento = "Mantenerlo abrigado, evitar corrientes de aire y jarabe específico."

        elif "rasca" in self.sintoma or "pulgas" in self.sintoma:
            self.diagnostico = "Dermatitis por picadura de pulgas"
            self.tratamiento = "Baño antipulgas, pipeta dermatológica y limpieza de su cama."

        elif "decaido" in self.sintoma or "no come" in self.sintoma:
            self.diagnostico = "Fiebre o dolor generalizado"
            self.tratamiento = "Medir temperatura, reposo absoluto y acudir a clínica urgente."

        else:
            self.diagnostico = "Condición inespecífica"
            self.tratamiento = "Observar 24 horas. Si empeora, llevar al hospital veterinario."


# --- FUNCIÓN PARA MOSTRAR LA TABLA DE PACIENTES ---
def mostrar_tabla_pacientes(lista_pacientes):
    if not lista_pacientes:
        print("\n[!] No hay pacientes registrados en el historial.")
        return

    # Encabezados de la tabla con su respectivo ancho para que quede alineado
    print("\n" + "="*115)
    print(f"{'MASCOTA':<12} | {'ESPECIE':<10} | {'EDAD':<6} | {'SÍNTOMA GENERAL':<20} | {'DIAGNÓSTICO IA':<32} | {'TRATAMIENTO SUGERIDO'}")
    print("="*115)

    # Imprimir cada paciente formateando el espacio
    for p in lista_pacientes:
        # Recortamos textos muy largos para que no rompan la tabla
        sintoma_corto = p.sintoma if len(p.sintoma) <= 20 else p.sintoma[:17] + "..."
        diag_corto = p.diagnostico if len(p.diagnostico) <= 32 else p.diagnostico[:29] + "..."

        print(f"{p.nombre_mascota:<12} | {p.especie:<10} | {p.edad:<6} | {sintoma_corto:<20} | {diag_corto:<32} | {p.tratamiento}")

    print("="*115 + "\n")


# --- PROGRAMA PRINCIPAL (MENÚ INTERACTIVO) ---
historial_veterinario = []

print("--------------------------------------------------")
print("  ¡Bienvenido al Sistema de IA Veterinaria v1.0! ")
print("--------------------------------------------------")

menu_activo = True

while menu_activo:
    print("--- MENÚ DE OPCIONES ---")
    print("1. Registrar mascota y consultar a la IA")
    print("2. Ver historial de pacientes (Tabla)")
    print("3. Salir del programa")

    opcion = input("Selecciona una opción (1-3): ")

    if opcion == "1":
        print("\n--- CONSULTA MÉDICA CON IA ---")
        nombre = input("Nombre de la mascota: ")
        especie = input("¿Qué animal es? (Perro, Gato, etc.): ")
        edad = input("Edad de la mascota (ej. 3 años / 5 meses): ")

        print("\nSíntomas comunes clave: [vomito/diarrea] [estornudo/tos] [rasca/pulgas] [decaido/no come]")
        sintoma = input("Describe el síntoma principal: ")

        # Creamos el objeto de la clase Veterinario
        nuevo_paciente = Veterinario(nombre, especie, edad, sintoma)

        # Ejecutamos la función de IA para procesar el diagnóstico
        nuevo_paciente.simular_ia_diagnostico()

        # Mostramos el resultado inmediato en pantalla
        print("\n" + "~"*50)
        print(f"  RESULTADO DEL ANÁLISIS DE LA IA PARA {nuevo_paciente.nombre_mascota.upper()}")
        print("~"*50)
        print(f" Diagnóstico Presuntivo: {nuevo_paciente.diagnostico}")
        print(f" Tratamiento Sugerido:   {nuevo_paciente.tratamiento}")
        print("~"*50)
        print("[✔] Paciente guardado con éxito en el historial.\n")

        # Guardamos el objeto en nuestra lista/historial
        historial_veterinario.append(nuevo_paciente)

    elif opcion == "2":
        mostrar_tabla_pacientes(historial_veterinario)

    elif opcion == "3":
        print("\nGracias por usar la IA Veterinaria. ¡Cuida mucho a tus mascotas! Bye.")
        menu_activo = False

    else:
        print("\n[!] Opción no válida. Por favor, selecciona 1, 2 o 3.\n")
```
